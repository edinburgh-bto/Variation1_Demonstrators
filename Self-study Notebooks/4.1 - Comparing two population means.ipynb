Last week we introduced inferential statistics: the collection of methods used to make conclusions about one or more populations based on small samples drawn from them. In science we are interested in understanding the properties of, and the differences between, populations. We are not interested in the specific small samples we collect from them in experimental and observational studies. This is because samples are affected by variation. For example, the mean beak depth of ten ground finches depends on which ten ground finches are sampled, whereas the mean beak depth of the population of all ground finches is just one value—and that is the value we want to know.

Last week we looked at one type of inference: the estimation of a population mean and its precision. This week we'll look at another type of inference: the comparison of the means of two populations to examine the evidence against them being the same.

The commonest method of comparing population means in the biological, geological and medical sciences is called **significance testing**. It is not the only method, and statisticians have been arguing about its pros and cons since its development 100 years ago. But it is the method most scientists have been taught and have used for the last century as it is fairly easy to apply and understand, especially with specialised computer software. It is the method you will see most often in published scientific papers. So it is the one method you need to know about. We do not have time in this course to introduce other statistical methods.

To develop the ideas behind significance testing we will use an example of the evolution of beak depths of medium ground finches.

## The evolution of medium ground finch beak depths on the Galápagos islands

Peter and Rosemary Grant and their colleagues have studied Galápagos finch species every year since 1976. One such species is the medium ground finch on the Galápagos island of Daphne Major.

The medium ground finch feeds on seeds. The finches have inherited variation in beak shape with some individuals having wide, deep beaks and others having thinner, shallower beaks. Large-beaked birds feed more efficiently on large, hard seeds, whereas smaller-beaked birds feed more efficiently on small, soft seeds. 

During 1977 a drought altered vegetation on the island. After the drought the number of seeds declined dramatically. The decline was greater in small, soft seeds than in large, hard seeds. Therefore, the Grants and colleagues hypothesised that large-beaked birds were better able to survive the drought than the small-beaked birds. 

As beak size is inherited from parent to offspring the theory of natural selection predicts that the drought would cause selection for finches with larger beaks. In other words, if the Grants were correct we might expect that beak sizes were larger in 1978 than 1976 on average.

We can test this hypothesis by comparing mean beak sizes of medium ground finches on Daphne Major before (1976) and after (1978) the drought.

The beak depth data collected by the Grants are in the file `Datasets/finches beak depth.csv`. We looked at the 1976 beak depths last week. Now we want to compare beak depths from both years.

## First plot the data

First, let's look at the beak depth data the Grants collected.

The file `finches beak depth.csv` contains the beak depths from 1976 and 1978.

Run the following code cell to plot frequency distributions (histograms) of medium ground finch beak depths sampled from Daphne Major in 1976 and 1978.

## Significance testing

Let's repeat our question. 

Is the observed mean difference of half a millimetre between 1976 and 1978 of biological or evolutionary interest, or is it just noise due to the random sampling of these specific finches?

Unfortunately, nothing, not even statistics, can provide a definitive answer to this question. We cannot say with absolute, 100% certainty either way if the difference is of biological or evolutionary interest or not. The reason for this is because the Grants only measured a small subset of finches in both years. As we saw last week, this implies that there is some uncertainty, some error, in the population mean beak depths for both years. Which implies there must be some uncertainty in the difference between those population means.

So we have to reformulate our question. Exactly how you reformulate the question depends on which branch of statistics you adhere to (of which there are four). Here we are learning about classical statistics, also called frequentist statistics. This is the type of statistics most scientists use most of the time.

In classical statistics we first set up what's called a **null hypothesis**. Null hypotheses can be thought of as "boring" hypotheses: the hypotheses of no difference, or no change, or no effect or no correlation. 

So our null hypothesis will be that population mean beak depths are the same in 1976 and 1978.

We then ask the following question:

<div class="alert alert-success">

If the null hypothesis were true, how likely are the data?

</div> 

In classical statistics, this is how you should **always** formulate your statistical questions. Another way of saying it is "How likely are my data under the null hypothesis?".

In terms of the ground finches we are asking, "If population mean beak depths are the same in 1976 and 1978, how likely (probable) would an observed difference of at least half a millimetre be?".

The answer to this question is a probability; a number between 0 and 1. If this probability is close to 1 then we can be fairly certain that the data are consistent with the null hypothesis. But never 100% certain. Conversely, if this probability is close to 0 then we can be fairly certain that the data are inconsistent with the null hypothesis, but again not 100% certain. This probability is usually given the symbol *P*. It's value is found by inputting the data into some statistical software. More on that in a moment.

The best way to interpret *P* is as the **strength of evidence against the null hypothesis**. The lower the value of *P* the stronger the evidence against the null hypothesis. 

The following figure is a reasonable guide on how to interpret the value of *P*.

<div>
<img src="Images/p_evidential_descriptors.png" width='20%'/>
</div>

This guide is not set in stone. It has no scientific justification. It is just a useful way of thinking about how strong the evidence is against the null hypothesis.

For example, if, after doing a significance test, you get a value of *P* above 0.1 (a probability above 10%) then you have insufficient evidence against the null hypothesis; your data are consistent with the null hypothesis being true. That does not mean that it is true; the null hypothesis may not be true and you just happened to get results that fooled you into thinking it was true. You can never be 100% certain about the results of a scientific experiment.

On the other hand, if you get a value of *P* less than 0.001 (a probability less than one in a thousand) then you have very strong evidence against the null hypothesis; your data are inconsistent with the null hypothesis being true. But again, there is always some amount of uncertainty. The null hypothesis may in fact be true and you just happened to get results that fooled you into thinking it was not true.

## Two sample *t*-test

The type of significance test you perform on your data depends on the nature of your data (e.g., categorical, numerical) and what you want to infer about the population it was sampled from (e.g., difference in means, correlation, regression, etc.). One of the problems students find difficult is knowing which test to use on their data. We won't worry about that in this course. We will tell you which significance test to use for your data.

The significance test comparing the means of two populations is called a **two sample *t*-test**. The "two sample" part is because we have two samples from two populations (1976 and 1978). The "*t*" part is the particular test-statistic used. We don't need to understand the details of how its value is calculated in the test. Other than to say it is a measure of the ratio of signal to noise in the data. The larger the difference in the means the larger the signal and the larger the value of *t*. The less variable the data are the less noise there is and the larger the value of *t*. Clearly a large signal-to-noise ratio is better, so a larger value of *t* is better.

To perform a two sample *t*-test in Python we need to import the code that performs the test from the `scipy.stats` module.

<div class="alert alert-info">

1. Read the code so that you understand what it does.
2. Run the code cell to perform the *t*-test.
</div>

```python
from scipy.stats import ttest_ind # import the two sample t-test function

# Perform the two sample t-test on the beak depth data
# Include nan_policy='omit' because there are missing values in the 1978 data that need to be ignored.
# Without this the test won't work.
test = ttest_ind( beak_depth['1978'], beak_depth['1976'], nan_policy='omit' )

# Print degrees of freedom
print(f'df = {test.df}')

# Print t to 2 sig fig
print(f't = {test.statistic:.2g}')

# Print P to 2 sig fig
print(f'P = {test.pvalue:.2g}')

# Print 95% CI of the difference in means
ci = test.confidence_interval()
print(f'95% CI = [{ci.low:.2g}, {ci.high:.2g}]')
```

The two sample *t*-test is performed by the function `ttest_ind()`. The beak depths of the two samples are passed to the function and it returns the values of *df*, *t*, *P* and the 95% confidence interval of the estimate of the difference in the means. All of the hard calculations and complicated maths are done for you by the computer.

It's as simple as that.

What is *df*? *df* stands for "degrees of freedom". In a two sample *t*-test the degrees of freedom is the number of observations minus two. There are 75 finches from 1976, 60 from 1978, so *df* = 75 + 60 - 2 = 133. You don't really need to know what this means but it should always be reported when you write up the results of a significance test in a report or poster. It lets the reader know the total number of observations made.

We've printed the value of the *t* statistic. Not because it's useful, but because it should also always be reported when you write up the results of a significance test in a report or poster.

The value of *P* is 0.0019. That is strong evidence against the null hypothesis that population mean beak depths are the same in 1976 and 1978.

Finally `ttest_ind()` calculates the 95% confidence interval of the difference in the population mean estimates. As we said above, there is uncertainty (error) in the estimates of mean beak depths both in 1976 and 1978. So there must also be uncertainty (error) in the difference between these estimates. 

## Significance

During the development of the concepts of classical statistical tests in the 20<sup>th</sup> century, a value of *P* less than 0.05 came to be considered "statistically significant". The history behind this is complicated and driven by clashes of the strong personalities of the academics who founded these concepts.

*P* < 0.05 has no scientific justification. It is mere convention that modern-day scientists use to classify research into "statistically significant" and "statistically non-significant". You will learn more about the history and consequences of this arbitrary threshold in  2<sup>nd</sup> year. 

But for now all you need to know is that when *P* < 0.05 you have a statistically significant result, and when P > 0.05 you have a statistically non-significant result.

## Reporting the result of the test

Having performed the *t*-test it needs to be reported. When reporting a *t*-test you should give
1. the estimated difference in the means (including the 95% CI or SEM),
2. whether you found a significant or non-significant difference,
3. the value of the test statistic and the degrees of freedom,
4. the value of *P*.

All of this information is needed so that readers know how large the difference is between means (with error), which statistical test you used, and the results of that test.

Below is an example of how you might report this test. **You should use something similar when you write up your ladybird size report.**

## Common misinterpretations of *P*

The correct definition of *P* is:

<div class="alert alert-success">

the probability of obtaining a test statistic that is at least as unusual as the one you observed **if the null hypothesis were true**. 

</div>

As you can imagine, understanding what this definition means is not easy. Which is why *P* is one of the most misunderstood concepts in statistics; by students, researchers and lecturers alike (even lecturers who teach statistics).

A good way to think about *P* is as the surprisingness or unusualness of data: the smaller the value *P*, the more surprising are the data compared to what the null hypothesis predicts. In other words, the strength of evidence against the null hypothesis.

The most common misinterpretation of the *P* is that it is the probability of the null hypothesis being true. It is not. For the simple reason that *P* is calculated **assuming** the null hypothesis is already true. 

Other misinterpretations occur when *P* is just above significance, say 0.06. Although you maybe tempted to, you cannot say any of the following:

1. The collection of more data will make the result significant (i.e., *P* will drop below 0.05).
    
    - This comment is often seen in final year Honours project reports. In fact, there is no guarantee that more data will make a non-significant result significant. (Moreover, it is possible that more data may even make a significant result non-significant.)

2. There is a trend to significance.

    - This is sometimes seen in scientific papers. Again the suggestion is that more data will make a non-significant result significant.
